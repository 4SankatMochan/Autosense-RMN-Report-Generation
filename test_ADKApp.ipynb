```python
import vertexai
# PROJECT_ID = "acn-cda"
# LOCATION = "us-central1"
# STAGING_BUCKET = "gs://abt-genai"
def artifact_service_builder():
  from google.adk.artifacts import GcsArtifactService
  return GcsArtifactService(bucket_name="acn-cda-adk-staging123")
import os
os.environ["GOOGLE_GENAI_USE_VERTEXAI"] = "True"
os.environ["GOOGLE_CLOUD_PROJECT"] = "acn-cda"
os.environ["GOOGLE_CLOUD_LOCATION"] = "us-central1"
from vertexai.preview import reasoning_engines
from data_science.agent import root_agent as Main_Agent
app = reasoning_engines.AdkApp(
    agent=Main_Agent,
    artifact_service_builder= artifact_service_builder
)
```

```python
import pprint
session = app.create_session(user_id = "user3")
session_id = session.id
print(f"session_id: {session_id}")
track = 0
questions = ["What are the total Planned Spend, Actual Spend, Impressions, Views, Clicks, Conversions, and Revenue for each campaign type?",
"Calculate the monthly CTR, CVR, ROAS, CPA, AOV, Budget Utilization %, and Spend Efficiency percentage for each campaign type.",
"Retrieve the weekly Conversions, Revenue, and AOV for Campaign 01 (for weeks 1, 2, 3, and 4).",]
for que in questions:
    print(f"question: {que}")
    for event in app.stream_query(user_id="user3",session_id=session_id,message=que):
        print("--------")
        try:
            print(event["content"]["parts"][0]["text"])
        except:
            pass
        print("*******")
```
